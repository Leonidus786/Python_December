```python
from numpy import random
import numpy as np
```

```python
# print(dir(numpy))
```

```python
x = random.randint(100)
print(x)
```

```python
y = random.rand()
print(y)
```

```python
my_random_array = random.randint(1000, size=(5))

print(my_random_array)
```

```python
U = random.randint(100, size=(3, 5))

print(U)
```

```python
z = random.rand(3, 5)

print(z)
```

```python
f = random.choice([3, 5, 7, 9])

print(f)
```

```python
r = random.choice([3, 5, 7, 9], p=[0.1, 0.3, 0.6, 0.0], size=(100))

print(r)
```

```python
x = random.choice([3, 5, 7, 9], p=[0.1, 0.3, 0.6, 0.0], size=(3, 5))

print(x)
```

```python
x = random.choice([3, 5, 7, 9], p=[0.1, 0.3, 0.6, 0.0], size=(3, 5, 3))

print(x)
```

```python
arr = np.array([1, 2, 3, 4, 5])

random.shuffle(arr)

print(arr)
```

```python
arr = np.array([1, 2, 3, 4, 5])

print(random.permutation(arr))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.distplot([0, 1, 2, 3, 4, 5])

plt.show()
```

```python
sns.distplot([0, 1, 2, 3, 4, 5], hist=False)

plt.show()
```

```python
x = random.normal(size=(2, 3))

print(x)
```

```python
x = random.normal(loc=1, scale=2, size=(2, 3))

print(x)
```

```python
sns.distplot(random.normal(size=1000), hist=False)

plt.show()
```
